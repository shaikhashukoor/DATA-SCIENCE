```python
# Tuple of key-value pairs
t = (("a", 1), ("b", 2), ("c", 3))

d = {}

for key, value in t:
    d[key] = value

print("Dictionary:", d)
```

```python
list1 = list(map(int, input("Enter first list: ").split()))
list2 = list(map(int, input("Enter second list: ").split()))

found = False

for i in list1:
    if i in list2:
        found = True
        break

print(found)
```
