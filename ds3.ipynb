```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([0,1])
y=np.array([3,6])
plt.subplot(2,3,1)
plt.xlabel("x-axis",color='blue')
plt.ylabel("y-axis",color='blue')
plt.title("title1")
plt.scatter(x,y)


ypoints=np.array([4,2,5,2,4])
plt.subplot(2,3,2)
xpoints=np.array([2,5,1,5,2])
plt.plot(xpoints,marker="*",color="orange")
plt.plot(ypoints,linestyle="dotted",color='blue')


ypoints = np.array([3, 8, 1, 10])
plt.subplot(2,3,3)
plt.plot(ypoints, marker = 'o', ms = 5, mec = 'black',mfc='yellow')



x = np.array([80, 85, 90, 95, 100, ])
plt.subplot(2,3,4)
plt.plot(x)
plt.title("title2")
plt.grid()



x=np.array([5,7,8,17,2,9])
y=np.array([99,86,87,86,103,87])
plt.subplot(2,3,5)
colors=np.array(["red","green","blue","black","orange","purple"])
size=np.array([100,210,4,5,70,12])
plt.scatter(x,y, c=colors,alpha=0.5,s=size)



x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])
plt.subplot(2,3,6)
plt.bar(x, y, color = "red",width=0.1)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([20, 35, 25, 20])
mylabel=["25","35","20","20"]
myexplode=[0.2,0,0,0]
mycolor= ["black","hotpink","b","orange"]

plt.pie(y, labels = mylabel,startangle=90,explode=myexplode,shadow=True,colors=mycolor)
plt.legend()
plt.show() 
```
